```python
city=swiggy['City'].value_counts().head(10)
plt.figure(figsize=(10,5))
plt.bar(city.index, city.values)
plt.title("Top 10 Food City ")
plt.xlabel('city')
plt.ylabel('Number of dish')
plt.show()
```

```python
restaurant=swiggy['Restaurant Name'].value_counts().head(10)
plt.figure(figsize=(40,25))
plt.bar(restaurant.index,restaurant.values)
plt.title('Top 10 Restaurant by Number of Dishes ',fontsize=50)
plt.xlabel('Restaurant',fontsize=35)
plt.ylabel('Number of Dishes',fontsize=35)
plt.show()
```

```python
avg_cate=swiggy.groupby('Category')['Price (INR)'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(40,20))
plt.bar(avg_cate.index,avg_cate.values)
plt.title('Most expensive food categories',fontsize=50)
plt.xlabel('Category',fontsize=30)
plt.ylabel('Price',fontsize=30)
plt.show()
```

```python
plt.figure(figsize=(10,8))
plt.hist(swiggy['Price (INR)'],bins=30)
plt.title('Price distibution')
plt.xlabel('Price')
plt.ylabel('Number of Dishes')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(swiggy['Rating'], bins=10)
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(swiggy['Price (INR)'])
plt.title('Boxplot of Dish Prices')
plt.ylabel('Price (INR)')
plt.show()
```

```python
avg_price=swiggy.groupby('City')['Price (INR)'].mean().sort_values()
plt.figure(figsize=(12,8))
plt.barh(avg_price.index,avg_price.values)
plt.xlabel('City')
plt.ylabel('Price')
plt.show()
```

```python
top_rest=swiggy.groupby('Restaurant Name')['Rating'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(30,15))
plt.bar(top_rest.index,top_rest.values)
plt.title('Top 10 Highest Rated Restaurants',fontsize=30)
plt.xlabel('Restaurant Name',fontsize=30)
plt.ylabel('Rating',fontsize=30)
plt.show()
```

```python
bins = [0,100,200,300,500,1000]
labels = ['Very Low','Low','Medium','High','Premium']

swiggy['Price Category'] = pd.cut(swiggy['Price (INR)'], bins=bins, labels=labels)

price_cat = swiggy['Price Category'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(price_cat.values,
        labels=price_cat.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Price Category Distribution')
plt.show()
```
